```python
from sklearn.model_selection import StratifiedKFold
import pandas as pd
import numpy as np
```

```python
labels_path = "../dataset/classes.txt"
labels = []
with open(labels_path, "r") as f:
    labels = [line.strip() for line in f.readlines()]
```

```python
label_indices = [sorted(set(labels)).index(label) for label in labels]  # Konwertuj na indeksy klas
```

```python
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
```

```python
fold_indices = list(skf.split(np.zeros(len(label_indices)), label_indices))
```

```python
fold_data = []
for fold, (train_idx, test_idx) in enumerate(fold_indices):
    for idx in train_idx:
        fold_data.append({"index": idx, "fold": fold, "split": "train"})
    for idx in test_idx:
        fold_data.append({"index": idx, "fold": fold, "split": "test"})
```

```python
df_folds = pd.DataFrame(fold_data)
df_folds.to_csv("crossval_folds.csv", index=False)
print("Zapisano plik crossval_folds.csv")
```
