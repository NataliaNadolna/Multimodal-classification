```python
import os
import random
import shutil
```

```python
lyrics_folder = "../dataset/Lyrics"
labels_file = "../dataset/classes.txt"
image_folder = "../dataset/spectograms/mel_graphic"
```

```python
output_base = "../dataset"
```

```python

with open(labels_file, "r", encoding="utf-8") as f:
    all_labels = [line.strip() for line in f.readlines()]

# Get matching (id, lyric_file, image_file, label)
data = []
for idx in range(1, len(all_labels) + 1):
    lyrics_filename = f"{idx:03}.txt"
    lyrics_path = os.path.join(lyrics_folder, lyrics_filename)
    
    image_filename = f"{idx:03}_mel.png"
    image_path = os.path.join(image_folder, image_filename)

    if os.path.exists(lyrics_path) and os.path.exists(image_path):
        data.append((idx, lyrics_path, image_path, all_labels[idx - 1]))

print(f"Found {len(data)} valid data points.")
```

```python
random.shuffle(data)
n = len(data)
train_split = int(0.8 * n)
val_split = int(0.9 * n)

splits = {
    "train": data[:train_split],
    "val": data[train_split:val_split],
    "test": data[val_split:]
}
splits
```

```python
for split_name, split_data in splits.items():
    split_folder = os.path.join(output_base, split_name)
    lyrics_out = os.path.join(split_folder, "lyrics")
    images_out = os.path.join(split_folder, "spectograms")
    os.makedirs(lyrics_out, exist_ok=True)
    os.makedirs(images_out, exist_ok=True)

    labels_out = []

    for idx, lyric_path, image_path, label in split_data:
        shutil.copy2(lyric_path, os.path.join(lyrics_out, os.path.basename(lyric_path)))
        shutil.copy2(image_path, os.path.join(images_out, os.path.basename(image_path)))
        labels_out.append(label)

    with open(os.path.join(split_folder, "classes.txt"), "w", encoding="utf-8") as f:
        for label in labels_out:
            f.write(f"{label}\n")

print("Data successfully split and copied.")
```
